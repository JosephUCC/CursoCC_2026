Este sistema debe ser capaz de aplicar un porcentaje de propina al cliente por medio de un porcentaje que se calcula con el monto de la cuenta final.

## Entrada

- Nombre del cliente
- Monto total de la cuenta
- Porcentaje de la propina

## Proceso

Una vez que se piden los datos de entrada el sistema debe validar que los datos ingresados sean los correctos para evitar una caida. Posteriormente se procede con el calculo de la propina por medio de un porcentaje aplicado a la cuenta final.

## Salida esperada

El sistema debe mostrar finalmente un mensaje detallando su factura con el nombre del usuario, el monto final, la propina agregada y un mensaje de agradecimiento.

```python
monto_valido = False
porcentaje_valido = False
nombre = input('Ingrese su nombre: ')
try:
    monto = float(input('Ingrese el monto de la cuenta: '))
    if monto < 0:
      print('El monto no puede ser negativo')
    else:
      monto_valido = True
except ValueError:
    print('El monto debe ser un número')

try:
    porcentaje = float(input('Ingrese el porcentaje de propina: '))
    if porcentaje < 1:
      print('El porcentaje debe ser mayor a 0')
      exit()
    elif porcentaje > 100:
      print('El porcentaje no puede ser mayor a 100')
      exit()
    else:
      porcentaje_valido = True
except ValueError:
    print('El porcentaje debe ser un número')
    exit()

if monto_valido and porcentaje_valido:
  descuento = monto / 100
  monto_final = monto + (porcentaje * descuento)
  print(f'Monto final: ${monto_final}')
  print(f'Gracias por su compra {nombre}')
else:
  print('Error en los datos ingresados')
```

# Problema #2: Detector de número par/impar

## Problema

EL sistema debe ser capaz de solicitar un numero y este debe ser analizado para indicarle al usuario si es par o impar.

## Entrada

Numero entero y positivo

## Proceso

El proceso es validar si el numero recibido es par o impar por medio de una divicion para ver si el residuo es 0 o no.

## Salida esperada

El mensaje de salida debe ser el numero ingresado junto con la descripcion PAR/IMPAR.

```python
numeroValido = False
try:
  numero = int(input('Ingrese un numero: '))
  numeroValido = True
except ValueError:
  print('ERROR: Debe ingresar un numero entero')

if numeroValido:
  if numero % 2 == 0:
    print(f'El numero {numero} es PAR')
  else:
    print(f'El numero {numero} es IMPAR')
else:
  print('Lo sentimos no es posible analizar el numero')
```

# Problema #3: Clasificador de edad

## Problema

Este sistema debe ser capaz de recibir una edad y posteriormente analizarla para indicarle al usuario si la persona es un niño/a, adolecente, adulto o adulto mayor.

## Entrada

Edad actual del usuario.

## Proceso

Validar que el dato ingresado sea un numero entero y si es correcto por medio de un 'if' analizar el rango de edad actual del usuario.

## Salida esperada

Imprimir la edad ingresada junto con una descripcion indicando si el usuario es un nino/a, adolecente, adulto o adulto mayor.

```python
# Escriba la solución aquí
edad_valida = False
try:
  edad = int(input('Ingrese su edad: '))
  edad_valida = True
  if edad < 0:
    print('La edad debe ser mayor a 0')
except ValueError:
  print('La edad debe ser un numero entero')

if edad_valida:
  if edad < 12:
    print(f'Usted tiene {edad} años y es un niño/a')
  elif edad >= 12 and edad < 18:
    print(f'Usted tiene {edad} años y es un adolecente')
  elif edad >= 18 and edad < 65:
    print(f'Usted tiene {edad} años y es un adulto')
  else:
    print(f'Usted tiene {edad} años y es un adulto mayor')
```

# Problema #4: Validador de compra con descuento

## Problema

El sistema debe ser capaz de analizar si el cliente es frecuente o no, en caso de serlo debe aplicarse un descuento adicional y en caso de no serlo no se le aplica ningun descuento.

## Entrada

- Nombre del ciente
- Monto total
- Cliente frecuente
## Proceso

Validar que los datos ingresados sean los correctos, una vez que esto este bien se verifica si el usuario es un cliente frecuente lo cual le debe generar automaticamente un descuento adicional, caso contrario no se le aplica ningun descuento.

## Salida esperada

Se imprime el nombre del cliente junto con el monto final y el descuento aplicado por ser cliente frecuente.

## Pruebas manuales


```python
cliente = input('Ingrese su nombre: ')
monto_valido = False
try:
  monto = float(input('Ingrese el monto de la compra: '))
  if monto > 0 and monto < 100000:
    monto_valido = True
  elif monto < 0:
    print('El monto no puede ser negativo')
except ValueError:
  print('El monto debe ser un numero')

try:
  cliente_frecuente = input('Es cliente frecuente? (si/no): ')
  if cliente_frecuente.lower().strip() == 'si':
    cliente_frecuente = True
  elif cliente_frecuente.lower() == 'no':
    cliente_frecuente = False
  else:
    print('Debe ingresar si o no')
except ValueError:
  print('Debe ingresar si o no')

if monto_valido and cliente_frecuente:
    descuento = monto * 0.10
    monto_final = monto - descuento
    print('---Detalle de facturacion---')
    print(f'Monto total: ${monto}')
    print(f'Descuento: ${descuento}')
    print(f'Monto final: ${monto_final}')
    print(f'Gracias por su compra {cliente}!')
elif monto_valido and not cliente_frecuente:
    print(f'Monto final: ${monto}')
    print(f'Gracias por su compra {cliente}')
else:
    print('No es posible ayudarle sin un monto.')
```
